```python

import pandas as pd

# A new entry starts with:
# Since 2018 :
# Since Apr 2025 :
# 2016 - 2018 :
# 2024 :
# 2018-2019 :

start_pattern = r'(?:Since\s+(?:(?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)\s+)?\d{4}|\d{4}(?:\s*[-–]\s*\d{4})?)\s*:'

rows = []

for idx, text in df_clean["Career Highlights"].dropna().items():

    
    matches = list(re.finditer(start_pattern, text))

    for i, match in enumerate(matches):

        
        start = match.end()

        
        if i + 1 < len(matches):
            end = matches[i + 1].start()
        else:
            end = len(text)

        career = text[start:end].strip()

        rows.append({
            "Name": df.loc[idx, "Name"],
            "Period": match.group().replace(" :", ":"),
            "Career": career
        })

career_extracted = pd.DataFrame(rows)

print(career_extracted.to_string())
```

```python

career_summary = df_clean[["Name"]].copy()

phd_mask = career_extracted["Career"].str.contains(
    r"\bPh\.?\s*D\b|\bPhD\b",
    case=False,
    regex=True,
    na=False
)


phd_year = (
    career_extracted[phd_mask]
    .groupby("Name")["Period"]
    .apply(lambda x: " | ".join(x))
    .rename("PhD Year")
)

def get_phd_institution(text):
    text = re.sub(
        r"\bPh\.?\s*D\b\.?",
        "",
        text,
        flags=re.IGNORECASE
    )

    text = re.sub(
        r"\b(in|of)\s+(Physics|Biology|Chemistry|Mathematics|Science|Engineering)\b",
        "",
        text,
        flags=re.IGNORECASE
    )

    return text.strip(" ,.-")


phd_institution = (
    career_extracted[phd_mask]
    .assign(
        Institution=lambda x:
        x["Career"].apply(get_phd_institution)
    )
    .groupby("Name")["Institution"]
    .apply(lambda x: " | ".join(x))
    .rename("PhD Institution")
)


def classify_position(text):

    text_lower = text.lower()

    if "postdoctoral" in text_lower or "post-doctoral" in text_lower:
        return "Postdoc"

    elif "assistant professor" in text_lower:
        return "Assistant Professor"

    elif "associate professor" in text_lower:
        return "Associate Professor"

    elif re.search(r"\bprofessor\b", text_lower):
        return "Professor"

    else:
        return "Other positions"


career_extracted["Position Type"] = (
    career_extracted.apply(
        lambda row: (
            "PhD"
            if phd_mask.loc[row.name]
            else classify_position(row["Career"])
        ),
        axis=1
    )
)

def combine_entries(group):

    return " | ".join(
        group["Period"] + " " + group["Career"]
    )
postdoc = (
    career_extracted[
        career_extracted["Position Type"] == "Postdoc"
    ]
    .groupby("Name")
    .apply(combine_entries, include_groups=False)
    .rename("Postdoc")
)

assistant_prof = (
    career_extracted[
        career_extracted["Position Type"] == "Assistant Professor"
    ]
    .groupby("Name")
    .apply(combine_entries, include_groups=False)
    .rename("Assistant Professor")
)

associate_prof = (
    career_extracted[
        career_extracted["Position Type"] == "Associate Professor"
    ]
    .groupby("Name")
    .apply(combine_entries, include_groups=False)
    .rename("Associate Professor")
)

professor = (
    career_extracted[
        career_extracted["Position Type"] == "Professor"
    ]
    .groupby("Name")
    .apply(combine_entries, include_groups=False)
    .rename("Professor")
)

other_positions = (
    career_extracted[
        career_extracted["Position Type"] == "Other positions"
    ]
    .groupby("Name")
    .apply(combine_entries, include_groups=False)
    .rename("Other positions")
)

career_summary = (
    career_summary
    .merge(phd_year, on="Name", how="left")
    .merge(phd_institution, on="Name", how="left")
    .merge(postdoc, on="Name", how="left")
    .merge(assistant_prof, on="Name", how="left")
    .merge(associate_prof, on="Name", how="left")
    .merge(professor, on="Name", how="left")
    .merge(other_positions, on="Name", how="left")
)


print(career_summary.to_string())
```
