Turning into SQL

```python
import sqlite3
```

```python
conn = sqlite3.connect("iisermohali_faculty.db")
```

```python
df_raw = pd.read_csv("iisermohali_faculty_complete_raw.csv")

df_raw.to_sql("faculty_raw", conn, if_exists="replace", index=False)
df_clean.to_sql("faculty_clean", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    (SELECT COUNT(*) FROM faculty_raw) AS raw_count,
    (SELECT COUNT(*) FROM faculty_clean) AS clean_count;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT Department, COUNT(*) AS Count
FROM faculty_clean
GROUP BY Department
ORDER BY Count DESC;
"""

sql_department_counts = pd.read_sql(query, conn)

sql_department_counts
```

```python
pandas_department_counts = (
    df_clean["Department"]
    .value_counts()
    .rename_axis("Department")
    .reset_index(name="Count")
)

pandas_department_counts
```
