Lets scrape the data

```python
faculty_data = []

for category_link , tab in zip(tab_names, tabs):
    category = category_link.get_text(strip=True)

    for table in tab.find_all("table"):
        header_row = table.find("tr")
        headers = [
            cell.get_text(" ",strip=True)
            for cell in header_row.find_all(["th", "td"])
        ]
        name_index = headers.index("Name") if "Name" in headers else None
        department_index = headers.index("Department") if "Department" in headers else None
        designation_index = headers.index("Designation") if "Designation" in headers else None
        email_index = headers.index("Email") if "Email" in headers else None

        #Data rows
        for row in table.find_all("tr")[1:]:
         cells = row.find_all("td")
        if not cells or name_index is None:
            continue
        name = cells[name_index].get_text(" ", strip=True)
        department = (
                cells[department_index].get_text(" ", strip=True)
                if department_index is not None
                else None
            )

        designation = (
                cells[designation_index].get_text(" ", strip=True)
                if designation_index is not None
                else None
            )

        email = (
                cells[email_index].get_text(" ", strip=True)
                if email_index is not None
                else None
            )
        link = cells[name_index].find("a")

        profile_link = link.get("href") if link else None
        facultydata.append({
                "Category": category,
                "Name": name,
                "Department": department,
                "Designation": designation,
                "Email": email,
                "Profile Link": profile_link
            })

df = pd.DataFrame(facultydata)

df.head()

```

```python
from urllib.parse import urljoin

base_url = "https://www.iisermohali.ac.in"

def fix_profile_link(x):

    if pd.isna(x):
        return None

    x = str(x)

    # Find the actual IISER faculty path
    if "/faculty/" in x:
        x = x[x.find("/faculty/"):]

    return urljoin(base_url, x)


df["Profile Link"] = df["Profile Link"].apply(fix_profile_link)
```

```python
def get_section_text(soup, heading_text):

    heading = soup.find(
        lambda tag: tag.name in ["h1", "h2", "h3", "h4"]
        and tag.get_text(" ", strip=True) == heading_text
    )

    if not heading:
        return None

    text = []

    for element in heading.find_all_next():

        if element.name in ["h1", "h2", "h3", "h4"]:
            break

        value = element.get_text(" ", strip=True)

        if value:
            text.append(value)

    return " ".join(text)
```

```python
def get_tab_text(soup, tab_name):

    tabs = soup.find_all(
        "a",
        attrs={"data-toggle": "sppb-tab"}
    )

    for tab in tabs:

        if tab.get_text(" ", strip=True).rstrip(":") == tab_name:

            target_id = tab.get("href")

            if not target_id:
                return None

            # Remove the # from the href
            target_id = target_id.lstrip("#")

            # Find the content directly by ID
            content = soup.find(id=target_id)

            if content:
                return content.get_text(" ", strip=True)

            return None

    return None
```

```python
def get_section_text(soup, heading_text):

    # First: try finding it as a tab
    tab = soup.find(
        "a",
        attrs={"data-toggle": "sppb-tab"},
        string=lambda x: x and x.strip().rstrip(":") == heading_text
    )

    if tab:
        target_id = tab.get("href")

        if target_id:
            content = soup.select_one(target_id)

            if content:
                return content.get_text(" ", strip=True)

    # Second: try finding it as a heading
    heading = soup.find(
        lambda tag:
        tag.name in ["h1", "h2", "h3", "h4"]
        and tag.get_text(" ", strip=True).rstrip(":") == heading_text
    )

    if heading:
        parts = []

        for element in heading.find_all_next():

            if (
                element.name in ["h1", "h2", "h3", "h4"]
                and element != heading
            ):
                break

            text = element.get_text(" ", strip=True)

            if text and text.rstrip(":") != heading_text:
                parts.append(text)

        return " ".join(parts).strip() or None

    return None
```

```python
import time

tab_map = {
    "Research Area": "Research Area",
    "Research Interests": "Research Interests",
    "Career Highlights": "Career Highlights",
    "Awards": "Awards & Recognitions",
    "Publications": "Publications"
}

# Create/reset the columns
for column in tab_map:
    df[column] = None

# Visit every profile
for i, row in df.iterrows():

    profile_url = row["Profile Link"]

    if pd.isna(profile_url):
        continue

    response = requests.get(profile_url)
    profile_soup = BeautifulSoup(response.text, "html.parser")

    for column, website_tab in tab_map.items():

        df.loc[i, column] = get_section_text(
            profile_soup,
            website_tab
        )

    print(i, row["Name"])

    time.sleep(0.3)

print("DONE")








```
