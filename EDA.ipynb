PhD year × department

```python
df["PhD Start Year"] = (
    df["PhD Year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
    .astype(float)
)

df["PhD Decade"] = (
    (df["PhD Start Year"] // 10 * 10)
    .astype("Int64")
    .astype(str)
    + "s"
)

phd_department = pd.crosstab(
    df["Department"],
    df["PhD Decade"]
)

print(phd_department)
```

Postdoc experience by department

```python
pd.crosstab(
    df["Department"],
    df["Postdoc"].notna()
)
```

Previous Career Experience

```python
df["Previous Career Experience"] = (
    df[[
        "Postdoc",
        "Assistant Professor",
        "Associate Professor",
        "Professor",
        "Other positions"
    ]]
    .fillna("")
    .astype(str)
    .apply(
        lambda row: "Yes" if any(x.strip() for x in row) else "No",
        axis=1
    )
)

Experience = pd.crosstab(
    df["Department"],
    df["Previous Career Experience"]
)
print(Experience)
```

Profile Completeness Score

```python
cols = [
    "Email",
    "Profile Link",
    "Research Area",
    "Research Interests",
    "Career Highlights",
    "Awards",
    "Publications",
    "PhD Institution"
]

df["Profile Completeness"] = df[cols].notna().sum(axis=1)

print(df["Profile Completeness"].describe())
```

```python
print(
    df.groupby("Department")["Profile Completeness"]
      .mean()
      .sort_values(ascending=False)
)
```

Text-length analysis

```python
df["Research Interests Length"] = df["Research Interests"].fillna("").str.len()
df["Career Highlights Length"] = df["Career Highlights"].fillna("").str.len()
df["Publications Length"] = df["Publications"].fillna("").str.len()
df[
    ["Research Interests Length",
     "Career Highlights Length",
     "Publications Length"]
].describe()
```
