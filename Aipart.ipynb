```python
import pandas as pd
```

```python
df = pd.read_csv(r"E:\SEM 5\IDC409\faculty_final.csv")
df
```

```python
df["Research Text"]= (
    df["Research Area"].fillna("") + " " +
    df["Research Interests"].fillna("")
)
print(df[["Name", "Research Text"]].head())
```

```python
import re
def clean_text(text):
    text= text.lower()
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
df["Research Text Cleaned"] = (
    df["Research Text"].apply(clean_text)
)        
print(df[["Name","Research Text Cleaned"]]) 
                        
```
